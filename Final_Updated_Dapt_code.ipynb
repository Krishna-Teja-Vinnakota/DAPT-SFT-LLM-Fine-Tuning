```python
from unsloth import FastLanguageModel
import torch

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Llama-3.2-3B-Instruct-bnb-4bit",
    max_seq_length=2048,
    load_in_4bit=True,
)
print("Model loaded")
```

```python
TEST_QUERIES = [
    "What are the two metro corridors proposed in the Agra Metro DPR and what are their lengths?",
    "What is the design PHPDT capacity and train composition proposed for Agra Metro?",
]

def generate(prompt, max_new_tokens=256):
    FastLanguageModel.for_inference(model)
    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
    with torch.no_grad():
        out = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            temperature=0.7,
            do_sample=True,
            repetition_penalty=1.5,
            pad_token_id=tokenizer.eos_token_id,
        )
    return tokenizer.decode(out[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True)

# Run test queries
for i, query in enumerate(TEST_QUERIES, 1):
    print(f"\n{'='*60}")
    print(f"Q{i}: {query}")
    print(f"{'='*60}")
    print(generate(query))
```

```python
print("=" * 60)
print("BEFORE FINE-TUNING")
print("=" * 60)

base_responses = []
for i, q in enumerate(TEST_QUERIES):
    print(f"\nQ{i+1}: {q}")
    r = generate(q)
    base_responses.append(r)
    print(f"A: {r}\n")
```

```python
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj",
                     "gate_proj", "up_proj", "down_proj"],
    lora_alpha=16,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=42,
)

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total = sum(p.numel() for p in model.parameters())
print(f"Trainable: {trainable:,} ({trainable/total*100:.2f}%) of {total:,}")
```

```python
from google.colab import files

print("Select ALL 5 DPR files at once in the upload dialog")
print("(Hold Ctrl on Windows / Cmd on Mac to select multiple files)\n")

uploaded = files.upload()  # Select all 5 files in ONE upload

all_text = ""
file_stats = []

for filename, content in uploaded.items():
    text = content.decode("utf-8", errors="replace")
    words = len(text.split())
    all_text += "\n\n" + text
    file_stats.append((filename, words))
    print(f"  ✓ {filename:<40} {words:,} words")

print(f"\nTotal files uploaded : {len(uploaded)}")
print(f"Total words          : {len(all_text.split()):,}")
print(f"Approx total tokens  : ~{int(len(all_text.split()) * 1.3):,}")

if len(uploaded) < 5:
    print(f"\n⚠️  WARNING: Only {len(uploaded)} file(s) uploaded — expected 5")
    print("Do NOT run this cell again. Re-run it and select all 5 files together.")
```

```python
from datasets import Dataset

all_chunks = []

for filename, content in uploaded.items():
    text = content.decode("utf-8", errors="replace")

    # Tokenize this DPR individually
    tokens = tokenizer.encode(text, add_special_tokens=False)

    # Chunk into 2048-token pieces with 128-token overlap
    stride = 2048 - 128
    file_chunks = []
    for i in range(0, len(tokens), stride):
        chunk = tokens[i : i + 2048]
        if len(chunk) < 64:
            break
        file_chunks.append({
            "text": tokenizer.decode(chunk, skip_special_tokens=True),
            "source": filename   # Track which DPR each chunk came from
        })

    all_chunks.extend(file_chunks)
    print(f"  {filename:<40} → {len(tokens):,} tokens → {len(file_chunks)} chunks")

# Shuffle so model sees all DPRs mixed, not one after another
import random
random.seed(42)
random.shuffle(all_chunks)

dataset = Dataset.from_list(all_chunks)
print(f"\nTotal chunks across all DPRs : {len(all_chunks)}")
print(f"Dataset ready for DAPT training")
```

```python
from trl import SFTTrainer
from transformers import TrainingArguments

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset,
    dataset_text_field="text",
    max_seq_length=2048,
    packing=True,
    args=TrainingArguments(
        num_train_epochs=2,
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        learning_rate=2e-4,
        warmup_steps=10,
        lr_scheduler_type="cosine",
        fp16=not torch.cuda.is_bf16_supported(),
        bf16=torch.cuda.is_bf16_supported(),
        optim="adamw_8bit",
        weight_decay=0.01,
        logging_steps=10,
        output_dir="./dpr_dapt_output",
        save_strategy="epoch",
        seed=42,
    ),
)

stats = trainer.train()
print(f"\nDone! Loss: {stats.training_loss:.4f} | Time: {stats.metrics['train_runtime']:.0f}s")
```

```python
print("=" * 60)
print("AFTER FINE-TUNING")
print("=" * 60)

ft_responses = []
for i, q in enumerate(TEST_QUERIES):
    print(f"\nQ{i+1}: {q}")
    r = generate(q)
    ft_responses.append(r)
    print(f"A: {r}\n")
```
